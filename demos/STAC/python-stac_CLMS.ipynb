# Access the CLMS data in the eodc stac

This notebook is a short overview on how to access the CLMS VPP data and seasonal Trajectories in the eodc stac

```python
import numpy as np
import matplotlib.pyplot as plt
import xarray as xr
import rioxarray  # noqa: F401
from pyproj import Transformer
import pystac_client
```


## VPP data

Access the collection

```python
eodc_catalog = pystac_client.Client.open("https://stac.eodc.eu/api/v1")

# Search the collection
results = eodc_catalog.search(
    collections=["clms-vpp"],
    
    bbox=[15.419565, 48.435618, 15.420367, 48.436191],
    datetime="2017-01-01/2024-12-31",
)


# list all items and their assets in the collection
items = list(results.items())
print(f"Found {len(items)} items")
for item in items:
    print(item.id, item.assets.keys())
```

Access a specific item (parameter)

```python
item_length = next(
    eodc_catalog.search(collections=["clms-vpp"], ids=["LENGTH"]).items()
)
print(item_length.properties)

# Set the season
asset_length_s1 = item_length.assets['s1']

# get url
store_url_length = asset_length_s1.href
print(store_url_length)
print(asset_length_s1.extra_fields)
```

Open the dataset with xarray

```python
length = xr.open_zarr(
    store_url_length,
    **asset_length_s1.extra_fields["xarray:open_zarr_kwargs"]
)

print(length)
```

Select a bbox and plot the time series of all points inside

```python
# Reproject bbox WGS84 → EPSG:3035 
transformer = Transformer.from_crs("EPSG:4326", "EPSG:3035", always_xy=True)
# 14.3240105188116866,48.3117521133462304 : 14.3256663397323243,48.3128676765390423

lon_min, lat_min = 14.3240105188116866, 48.3117521133462304
lon_max, lat_max = 14.3256663397323243, 48.3128676765390423

x_min, y_min = transformer.transform(lon_min, lat_min)
x_max, y_max = transformer.transform(lon_max, lat_max)

print(f"EPSG:3035 bbox: x=({x_min:.0f}, {x_max:.0f}), y=({y_min:.0f}, {y_max:.0f})")
```

```python
length = length.assign_coords(time=length["data"].attrs["years"])

# Assign spatial coordinates from transform
transform = length["data"].attrs["transform"]
res_x, _, left, _, neg_res_y, top = transform
height, width = length["data"].shape[1], length["data"].shape[2]

xs = left     + res_x     * np.arange(width)  + res_x     / 2
ys = top      + neg_res_y * np.arange(height) + neg_res_y / 2

length = length.assign_coords(x=("x", xs), y=("y", ys))

# Subset to bbox 
subset = length["data"].sel(
    x=slice(x_min, x_max),
    y=slice(y_max, y_min)  # y is descending
)
print(f"Subset shape: {subset.shape}  ({subset.shape[1] * subset.shape[2]} pixels)")
```

```python
# Plot
fig, ax = plt.subplots(figsize=(10, 5))

years = subset.coords["time"].values

for row in range(subset.shape[1]):
    for col in range(subset.shape[2]):
        values = subset[:, row, col].values
        mask   = np.isfinite(values)
        if mask.any():
            ax.plot(
                years[mask], values[mask],
                marker="o", linewidth=1, alpha=0.7,
                label=f"({row},{col})"
            )

ax.set_title("Length Season 1 — per-pixel time series")
ax.set_xlabel("Year")
ax.set_ylabel("length value")
#ax.legend(loc="upper left", fontsize=8, ncol=2)
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

Select a different item

```python
item_EOSV = next(
    eodc_catalog.search(collections=["clms-vpp"], ids=["EOSV"]).items()
)
print(item_EOSV.properties)

# Set the season
asset_eosv_s1 = item_EOSV.assets['s1']

# get url
store_url_eosv = asset_eosv_s1.href
print(store_url_eosv)
print(asset_eosv_s1.extra_fields)
```

## Seasonal Trajectories

Have a look at the stac collection

```python

eodc_catalog = pystac_client.Client.open("https://stac.eodc.eu/api/v1")

# Search the collection
results = eodc_catalog.search(
    collections=["clms-st"],
    
    bbox=[15.419565, 48.435618, 15.420367, 48.436191],
    datetime="2017-01-01/2024-12-31",
)

items = list(results.items())
print(f"Found {len(items)} items")
for item in items:
    print(item.id, item.assets.keys())
```

Select Item and Asset

```python
item = next(
    eodc_catalog.search(collections=["clms-st"], ids=["PPI"]).items()
)
print(item.properties)

print(item.assets.keys())


asset = item.assets['data']
# get url
store_url = asset.href
print(store_url)
print(asset.extra_fields)
```

Access with xarray and assign time as a coordinate

```python
import xarray as xr
coords = xr.open_zarr(
    store_url,
    **asset.extra_fields["xarray:open_zarr_kwargs"]
)

print(coords)

coords = coords.assign_coords(time=coords["PPI"].attrs["times"])
```

Create a subset from the bbox from the VPP data

```python
transform = coords["PPI"].attrs["transform"]
res_x, _, left, _, neg_res_y, top = transform
height, width = coords["PPI"].shape[1], coords["PPI"].shape[2]

xs = left     + res_x     * np.arange(width)  + res_x     / 2
ys = top      + neg_res_y * np.arange(height) + neg_res_y / 2

coords = coords.assign_coords(x=("x", xs), y=("y", ys))

# ── Subset to bbox ────────────────────────────────────────────────────────────
subset = coords["PPI"].sel(
    x=slice(x_min, x_max),
    y=slice(y_max, y_min)  # y is descending
)
print(f"Subset shape: {subset.shape}  ({subset.shape[1] * subset.shape[2]} pixels)")
```

Plot a single time stamp

```python
TARGET_DATE = "2019-03-01T00:00:00"

# select just this one timestamp from the same PPI array
subset_t = coords["PPI"].sel(
    x=slice(x_min, x_max),
    y=slice(y_max, y_min),  # y is descending
).sel(time=TARGET_DATE)

print(f"Subset shape: {subset_t.shape}  ({subset_t.shape[0] * subset_t.shape[1]} pixels)")

# force the actual read now, so any 403/404 issue surfaces immediately
data = subset_t.values
print("min:", np.nanmin(data), "max:", np.nanmax(data), "finite pixels:", np.isfinite(data).sum())

fig, ax = plt.subplots(figsize=(8, 6))
im = ax.imshow(data, cmap="viridis")
ax.set_title(f"PPI — {TARGET_DATE}")
plt.colorbar(im, ax=ax, label="PPI")
plt.tight_layout()
plt.show()
```

Plot a time series

```python
subset_2024 = subset.sel(time=slice("2024-01-01T00:00:00", "2024-03-31T23:59:59"))
print(f"2024 subset shape: {subset_2024.shape}  ({subset_2024.shape[0]} timestamps)")

years = subset_2024.coords["time"].values

data_2024 = subset_2024.load()  

fig, ax = plt.subplots(figsize=(10, 5))
years = data_2024.coords["time"].values

for row in range(data_2024.shape[1]):
    for col in range(data_2024.shape[2]):
        values = data_2024[:, row, col].values  
        mask   = np.isfinite(values)
        if mask.any():
            ax.plot(
                years[mask], values[mask],
                marker="o", linewidth=1, alpha=0.7,
                label=f"({row},{col})"
            )

ax.set_title("PPI Season 1 — per-pixel time series")
ax.set_xlabel("Date")
ax.grid(True, alpha=0.3)
# ax.legend(loc="upper left", fontsize=8, ncol=2)
plt.tight_layout()
plt.show()
```
